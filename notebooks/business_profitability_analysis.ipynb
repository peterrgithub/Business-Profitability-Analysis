# Executive Summary

## Business Question

Management wants to understand why profitability varies across the business despite strong sales performance. This analysis examines overall profitability, category and regional differences, product-level losses, and the relationship between discounts and recorded profit.

## Key Findings

* The dataset contains **9,994 order-line records representing 5,009 unique orders**.
* Total recorded sales are approximately **USD 2.30 million**, with approximately **USD 286,397 in recorded profit**, giving an overall recorded profit margin of approximately **12.47%**.
* Profitability varies substantially by category. **Furniture** recorded a much lower aggregate profit margin than **Office Supplies** and **Technology**, despite generating substantial sales.
* Regional profitability also varies. The **West** region recorded the highest aggregate sales, profit, and recorded profit margin among the four regions, while the **Central** region recorded the lowest aggregate profit margin.
* At the product level, **301 of 1,850 unique products (16.27%) were loss-making overall**.
* These loss-making products generated approximately **USD 555,414 in sales** and a combined recorded loss of approximately **USD 77,068**, representing about **24.18% of total sales**.
* The ten products with the largest dollar losses accounted for approximately **38.22% of the total recorded loss among loss-making products**, indicating that losses were concentrated among a relatively small group of products.
* Discount levels show a strong descriptive association with profitability in the analyzed data. Higher discount levels generally correspond with increasingly negative recorded profit margins, particularly within products that are loss-making overall.
* However, the analysis **does not establish that discounts caused the losses**. Discount levels may be associated with product mix, sub-category, region, customer characteristics, or other factors.

## Management Investigation Areas

The results suggest several areas that management could investigate further:

1. Products generating substantial dollar losses should be reviewed individually.
2. Furniture profitability warrants investigation because of its comparatively low aggregate recorded margin.
3. High-discount transactions should be examined alongside product, customer, regional, and order-level characteristics rather than in isolation.
4. Products with very small sales volumes should be interpreted cautiously because extreme percentage margins can result from small denominators.
5. Any future analysis should preserve the dataset grain and validate joins, aggregations, and metric definitions before drawing conclusions.

## Analytical Limitation

This analysis is descriptive and based on the fields available in the Sample Superstore dataset. It identifies patterns and areas for investigation but does not establish causal relationships. In particular, the relationship between discount and profitability should not be interpreted as evidence that discounts independently caused the observed losses.

```python
import sqlite3

conn = sqlite3.connect("../data/superstore.db")

df.to_sql(
    "superstore",
    conn,
    if_exists="replace",
    index=False
)

print("Superstore data loaded into SQLite successfully.")
```

```python
query = """
SELECT
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit,
    ROUND(
        SUM(Profit) * 100.0 / SUM(Sales),
        2
    ) AS Profit_Margin_Percent
FROM superstore;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Category,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit,
    ROUND(
        SUM(Profit) * 100.0 / SUM(Sales),
        2
    ) AS Profit_Margin_Percent
FROM superstore
GROUP BY Category
ORDER BY Profit_Margin_Percent DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Region,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit,
    ROUND(
        SUM(Profit) * 100.0 / SUM(Sales),
        2
    ) AS Profit_Margin_Percent
FROM superstore
GROUP BY Region
ORDER BY Profit_Margin_Percent DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Discount,
    COUNT(*) AS Order_Line_Count,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit,
    ROUND(
        SUM(Profit) * 100.0 / SUM(Sales),
        2
    ) AS Profit_Margin_Percent
FROM superstore
GROUP BY Discount
ORDER BY Discount;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    "Product Name",
    COUNT(*) AS Order_Line_Count,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit,
    ROUND(
        SUM(Profit) * 100.0 / SUM(Sales),
        2
    ) AS Profit_Margin_Percent
FROM superstore
GROUP BY "Product Name"
HAVING SUM(Profit) < 0
ORDER BY Total_Profit ASC;
"""

pd.read_sql_query(query, conn)
```
